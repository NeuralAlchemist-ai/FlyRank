## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a scoring (Predictive Risk Scoring) applied as a ranking task. While I could treat this as a simple classification problem (like predicting "Will it drop? Yes or No"), just getting a "Yes" isn't super helpful if there are thousands of pages flagged. By treating this as a scoring task, the model will give each page a risk probability. Then, I can rank them from highest risk to lowest, giving the editorial team a clear, prioritized to-do list so they know exactly where to start.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

A future traffic drop (observed outcome). Instead of trying to predict a made-up score, I want to predict something real. My target will be whether a page actually drops in traffic (e.g., a >20% loss in clicks) during a future 30-day window. This is a true outcome based on actual historical data, not just the model copying an existing rule-based flag.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@50. Standard metrics like general accuracy or ROC AUC aren't the best fit here because a human team can only review a limited number of pages anyway. What really matters is that the top pages my model recommends are actually the ones in trouble. Precision@50 tells me exactly what percentage of my top 50 recommendations were actually correct, which directly measures how useful this tool would be in the real world.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Search trends are too messy for a simple if-statement. A fixed rule is way too reactive by the time the rule triggers, the traffic is already gone. A page might have stable traffic but slowly dropping click-through rates and a slipping average position. A machine learning model can weigh dozens of these small, interacting signals all at once to catch a downward momentum shift before the traffic completely crashes. A human-written rule just can't scale to do that reliably
