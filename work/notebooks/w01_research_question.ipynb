Freestyle Direction: Growth / Recovery / Momentum Prediction. I am choosing this freestyle lane because it closely mirrors the reality of production ML infrastructure. ## 2. The question: decision, action, cost of a wrong call

The Search Question: Can we predict which pages are highly likely to suffer a severe traffic decline (or experience a momentum breakout) in the next 30 days based on their trajectory over the prior 90 days?

Unit of Analysis: A specific page (content_hash_id) measured over a defined time window.

The Decision & Action: This improves content maintenance triage. If the model flags a high-value page as a high future-decline risk, a human editor takes action to proactively review, rewrite, or protect the page before the traffic loss solidifies.

Cost of a Wrong Call: A false positive wastes a reviewer's time on a page that was going to be fine. A false negative is an expensive missed opportunity, resulting in unmitigated traffic decay.

Why ML/Data: Simple heuristics are too noisy and reactive. A learned model can weigh complex interactions between search volume, position drift, and engagement decay to catch subtle momentum shifts earlier.

## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
declining_pct = (df["trend_direction"] == "down").mean() * 100
avg_days_since_update = df["days_since_last_update"].mean()

print(f"Total pages in starter sample: {total_pages:,}")
print(f"Pages currently labeled as declining: {declining_pct:.1f}%")
print(f"Average days since last update: {avg_days_since_update:.1f} days")
```

## 4. Careful words: what I can and can't claim

I can build an observational, decision-support tool. I can claim that certain historical signal patterns are probabilistically associated with future traffic drops, and I can rank pages based on that risk to optimize a team's workflow. However, I cannot claim causal proof and guarantee that rewriting a flagged page will inherently reverse the trend. Furthermore, I cannot claim that this model proves or "reverse-engineers" Google's core algorithm. It strictly observes historical correlation.
